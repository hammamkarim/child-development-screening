## Simpan ke CSV

```python
## Gabungkan data percentile

percentile_path = Path("../data/denver/denver_percentiles_source.csv")

df_percentiles = pd.read_csv(percentile_path)

PERCENTILE_COLUMNS = [
    "p25_month",
    "p50_month",
    "p75_month",
    "p90_month"
]

assert len(df_percentiles) == 125, (
    f"Jumlah data percentile harus 125, ditemukan {len(df_percentiles)}"
)

assert df_percentiles["id"].is_unique, (
    "ID pada file percentile tidak boleh duplikat."
)

assert set(df_percentiles["id"]) == set(df_items["id"]), (
    "ID percentile tidak sama dengan ID 125 item Denver II."
)

df_master = df_items.merge(
    df_percentiles,
    on="id",
    how="left",
    validate="one_to_one"
)

df_master = df_master[[
    "id",
    "domain",
    "name",
    "reportable",
    "footnote",
    "p25_month",
    "p50_month",
    "p75_month",
    "p90_month"
]]

print("✓ Data percentile berhasil digabungkan.")
print("Jumlah item:", len(df_master))
```

## validasi percentile

```python
## Validasi percentile

PERCENTILE_COLUMNS = [
    "p25_month",
    "p50_month",
    "p75_month",
    "p90_month"
]

# Semua nilai yang tersedia harus numerik
for col in PERCENTILE_COLUMNS:
    df_master[col] = pd.to_numeric(
        df_master[col],
        errors="coerce"
    )

# Validasi urutan percentile
for _, row in df_master.iterrows():

    values = [
        row["p25_month"],
        row["p50_month"],
        row["p75_month"],
        row["p90_month"]
    ]

    # Abaikan nilai yang memang tidak tersedia
    values = [v for v in values if pd.notna(v)]

    assert values == sorted(values), (
        f"Urutan percentile salah pada {row['id']}: {values}"
    )

print("✓ Validasi percentile selesai.")
print("✓ Semua nilai percentile yang tersedia berurutan.")
print("✓ Nilai kosong hanya diperbolehkan jika memang tidak tersedia.")
```
